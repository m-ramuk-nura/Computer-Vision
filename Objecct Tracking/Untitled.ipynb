```python
import cv2
```

```python
TrDict = {'csrt':cv2.legacy.TrackerCSRT_create,
          'kcf':cv2.TrackerKCF_create,
         'boosting':cv2.legacy.TrackerBoosting_create,
         'mil':cv2.TrackerMIL_create,
         'tld':cv2.legacy.TrackerTLD_create,
         'mediaflow':cv2.legacy.TrackerMedianFlow_create,
         'mosse':cv2.legacy.TrackerMOSSE_create}
```

```python
tracker = TrDict['csrt']()
```

```python
video = cv2.VideoCapture(r'mot.mp4')
```

```python
ret, frame = video.read()
```

```python
cv2.imshow('Frame',frame)
bb = cv2.selectROI('Frame',frame)
tracker.init(frame,bb)
```

```python
while True:
    ret, frame = video.read()
    if not ret:
        break
    (success,box)= tracker.update(frame)
 
    if success:
        (x,y,w,h) = [int(a) for a in box]
        cv2.rectangle(frame,(x,y),(x+w,y+h),(255,255,0),2)
    cv2.imshow('Frame',frame)
    key = cv2.waitKey(5) & 0xFF
    if key==ord('q'):
        break
video.release()
cv2.destroyAllWindows()
```

```python
trackers = cv2.legacy.MultiTracker_create()
ret, frame = video.read()
tracker = TrDict['boosting']()
video = cv2.VideoCapture(r'mot.mp4')
```

```python
k = 3

for i in range(k):
    cv2.imshow('Frame',frame)
    cv2.waitKey(1)
    bbi = cv2.selectROI('Frame',frame)
    tracker_i = TrDict['boosting']()
    trackers.add(tracker_i,frame,bbi)
```

```python
while True:
    ret, frame = video.read()
    if not ret:
        break
    (success,boxes)= trackers.update(frame)
 
    for box in boxes:
        (x,y,w,h) = [int(a) for a in box]
        cv2.rectangle(frame,(x,y),(x+w,y+h),(255,255,0),2)
    cv2.imshow('Frame',frame)
    key = cv2.waitKey(5) & 0xFF
    if key==ord('q'):
        break
video.release()
cv2.destroyAllWindows()
```
