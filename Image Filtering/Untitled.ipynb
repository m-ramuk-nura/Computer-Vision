```python

import numpy as np
import matplotlib.pyplot as plt
import cv2 
from scipy import signal
```

```python
im = cv2.imread(r'image.jpeg')
```

```python
im = im[:,:,::-1]
```

```python
plt.imshow(im)
```

```python
imgray = cv2.cvtColor(im,cv2.COLOR_RGB2GRAY)
```

```python
smoothingMask = np.ones((15,15))/255
```

```python
imBlurred = signal.convolve2d(imgray,smoothingMask,boundary='symm',mode='same')
```

```python

plt.imshow(imBlurred, cmap='grey')
```

```python
xMask = np.array([[-1,0,1],[-1,0,1],[-1,0,1]])
yMask = xMask.T.copy()
```

```python
fx = signal.convolve2d(imgray,xMask,boundary='symm',mode ='same')
fy = signal.convolve2d(imgray,yMask,boundary='symm',mode ='same')
```

```python
Gm = (fx**2 + fy**2)**0.7
```

```python
plt.imshow(fx,cmap='grey')
```

```python
plt.imshow(fy,cmap='grey')
```

```python
plt.imshow(Gm,cmap='grey')
```

```python
imsharped = imBlurred + Gm*0.1
```

```python
plt.imshow(imsharped,cmap='grey')
```
